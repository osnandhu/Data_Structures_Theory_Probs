TWO POINTER METHOD aka sliding window
 -> effectiveness and simplicity in sorted arrays
 prerequisite: NEED A SORTED ARRAY
 Synopsis:
 Imagine two people a.k.a "pointers" standing at the two ends of an array and trying to find a particular pair of elements that fulfill a specific condition.

references:
https://www.geeksforgeeks.org/two-pointers-technique/
https://mrmuhammadazeemrao.medium.com/the-two-pointers-technique-in-python-a-deep-dive-c556875565f2

Time complexity: 
In the worst-case scenario, the left pointer moves from the start to the end of the array and the right pointer moves from the end to the start, hence, making it linear in nature.

Time Complexity: O(N)

Why sorted arrays:
The two pointers technique, as illustrated, specifically relies on the fact that the array is sorted. This ensures that moving the right pointer leftward results in a smaller sum and moving the left pointer rightward gives a larger sum. If the array isn’t sorted, this assumption breaks down.





```python
# 11. Container with most water- https://leetcode.com/problems/container-with-most-water/description/?envType=problem-list-v2&envId=array&difficulty=MEDIUM
class Solution:
    def maxArea(self, height: List[int]) -> int:

        ## we need two pointers - initialise them
        left = 0
        right = len(height)-1
        max_area = 0

        ## to calculate the area, we need minimum of left height, right height* difference between the indices of left and right
        while left < right:
            area = min(height[left],height[right]) * (right - left)
            max_area = max(max_area,area)

            if height[left] < height[right]:
                left += 1
            else:
                right -=1
        return max_area  
```

```python
# 202 Happy Number https://leetcode.com/problems/happy-number/?envType=problem-list-v2&envId=two-pointers&difficulty=EASY
class Solution:
    def isHappy(self, n: int) -> bool:
        current_number = n
        sq_sum = 0
        numbers = {}
        while True:
            for i in str(current_number):
                sq_sum += int(i)**2
            if sq_sum == 1:
                return True
            if sq_sum in numbers:
                return False
            numbers[sq_sum] = 0
            current_number = sq_sum
            sq_sum = 0


        


```

```python
#1961. Check If String Is a Prefix of Array   https://leetcode.com/problems/check-if-string-is-a-prefix-of-array/description/?envType=problem-list-v2&envId=two-pointers&difficulty=EASY
class Solution:
    def isPrefixString(self, s: str, words: List[str]) -> bool:
        i = 0
        for word in words: 
            if s[i:i+len(word)] != word: return False 
            i += len(word)
            if i == len(s): return True 
        return False 
```
