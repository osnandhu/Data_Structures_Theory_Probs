```python
#contains duplicate or not: - easy
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        hashset = set()

        for n in nums:
            if n in hashset:
                return True
            hashset.add(n)
        return False
```

```python
# valid anagram - return true if anagram else return false - easy
from collections import Counter
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if (Counter(s) == Counter(t)):
            return True
        else:
            return False

        
        
```

```python
'''
EASY
Given an array of integers nums and an integer target, return the indices i and j such that nums[i] + nums[j] == target and i != j.

You may assume that every input has exactly one pair of indices i and j that satisfy the condition.

Return the answer with the smaller index first.
Input: 
nums = [3,4,5,6], target = 7

Output: [0,1]
Explanation: nums[0] + nums[1] == 7, so we return [0, 1].

Example 2:

Input: nums = [4,5,6], target = 10

Output: [0,2]


'''

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        final_list = []
        for i in range(0,len(nums)):
            temp = target - nums[i]
            if temp in nums:
                if(nums.index(temp)!= i):
                    final_list.append(i)
                    final_list.append(nums.index(temp))
                    break;
            else:
                pass
        return list(set(final_list))
                

        
```

```python
#palindrome:  https://neetcode.io/problems/is-palindrome
class Solution:
    def isPalindrome(self, s: str) -> bool:
        if(len(s) > 0):
            joined_string = (''.join(e for e in s if e.isalnum())).lower()
            print(joined_string)
            reverse_string = joined_string[::-1]
            if joined_string == reverse_string:
                return True
            else:
                return False
        else:
            return True

         

        
```

```python
#majority element:

from collections import Counter
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        return Counter(nums).most_common()[0][0]
```

```python
#search insert position: https://leetcode.com/problems/search-insert-position/description/?envType=problem-list-v2&envId=array&difficulty=EASY
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        if target not in nums:
            nums.append(target)
            return sorted(nums).index(target)
        else:
            return nums.index(target)
```
