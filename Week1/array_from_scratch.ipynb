```python

import array
arr = array.array('i',[1,2,3,4,5,6,7,8])
```

```python
for x in arr:
    print(x, end=" ")
```

```python
#searching in the array
def findElement(arr,key):
    for i in range(len(arr)): 
        if (arr[i] == key):
            return i
    return -1
```

```python
#find and delete element 
def deleteElement(arr,key):
    pos = findElement(arr,key)
    if pos == -1:
        print("Cant find the element to be deleted")
    else:
        for i in range(pos,len(arr)-1):
            arr[i] = arr[i+1]
    return arr[0:len(arr)-1]

```

```python
new_arr = deleteElement(arr,4)
print("the new array post deletion")
for i in range(0,len(new_arr)):
    print(new_arr[i],end = " ")
```

```python
#insertion in an array
def insertElement(arr,key,pos):
    for i in range(pos-1,len(arr)-1,-1):
        arr[i+1] = arr[i]
    arr[pos] = key
    return arr

array_i = insertElement(arr,9,3)
print(array_i)
print("array after insertion")
for i in range(0,len(array_i)):
    print(array_i[i],end = " ")
```
